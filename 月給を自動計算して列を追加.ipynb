```python
import pandas as pd

# 勤怠データのサンプルを作成
data = {
    "氏名": ["田中さん", "鈴木さん", "佐藤さん", "山田さん", "伊藤さん"],
    "出勤日数": [20, 18, 22, 15, 21],
    "残業時間": [10, 25, 5, 30, 8],
    "時給": [2000, 2000, 2500, 2000, 2200]
}

df = pd.DataFrame(data)

# 月給を自動計算して列を追加
df["月給"] = df["出勤日数"] * df["時給"] * 8  # 1日8時間勤務
df["残業代"] = df["残業時間"] * df["時給"] * 1.25  # 残業は1.25倍
df["支給合計"] = df["月給"] + df["残業代"]

# 結果を表示
# print(df)
# df　はprintの後に記載しないとダメ
print("---")
print(f"全員の支給合計：{df['支給合計'].sum():,}円")
print(f"一人あたり平均：{df['支給合計'].mean():,.0f}円")

df
```
