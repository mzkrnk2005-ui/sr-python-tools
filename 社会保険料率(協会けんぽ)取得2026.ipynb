```python
import requests
from bs4 import BeautifulSoup

# 令和8年度の保険料額表ページ
url = "https://www.kyoukaikenpo.or.jp/g7/cat330/sb3150/r08/r8ryougakuhyou3gatukara/"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}

response = requests.get(url, headers=headers)
print(f"ステータスコード：{response.status_code}")
print(f"ページタイトル確認：")

soup = BeautifulSoup(response.content, "html.parser")
title = soup.find("title")
print(title.text if title else "タイトルなし")
```

```python
# ページの構造を確認する
tables = soup.find_all("table")
print(f"テーブルの数：{len(tables)}個")

if tables:
    print("\n最初のテーブルの内容（先頭5行）：")
    rows = tables[0].find_all("tr")
    for row in rows[:5]:
        cells = row.find_all(["td", "th"])
        texts = [cell.get_text(strip=True) for cell in cells]
        print(texts)
```

```python
# ページのリンク一覧を確認する
# 都道府県別のPDFや別ページへのリンクがあるはず
links = soup.find_all("a", href=True)

print("ページ内のリンク一覧：")
for link in links:
    href = link.get("href", "")
    text = link.get_text(strip=True)
    # 保険料率に関連しそうなリンクだけ表示
    if any(keyword in text or keyword in href
           for keyword in ["保険料", "料率", "pdf", "PDF", "r8", "r07"]):
        print(f"  テキスト：{text}")
        print(f"  URL：{href}")
        print()
```

```python
import pandas as pd
import requests
import io

# 令和8年度 都道府県別保険料率のExcelを直接ダウンロード
url = "https://www.kyoukaikenpo.or.jp/assets/r8ippan3.xlsx"
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}

response = requests.get(url, headers=headers)
print(f"ステータスコード：{response.status_code}")
print(f"ファイルサイズ：{len(response.content):,} bytes")

# メモリ上でExcelを読み込む
excel_data = io.BytesIO(response.content)
df_raw = pd.read_excel(excel_data, sheet_name=None)  # 全シートを読み込む

print(f"\nシート一覧：")
for sheet_name in df_raw.keys():
    print(f"  ・{sheet_name}")
```

```python
# 東京の保険料率を抽出する
df_tokyo = df_raw["東京"]

# 6行目に料率が入っている（0始まりなのでindex=6）
rate_row = df_tokyo.iloc[6]
print("6行目の内容：")
print(rate_row.values)
print()

# 料率を取得
kenko_rate     = float(rate_row.iloc[5])   # 健康保険料率（介護なし）
kenko_kaigo    = float(rate_row.iloc[7])   # 健康保険料率（介護あり）
nenkin_rate    = float(rate_row.iloc[11])  # 厚生年金保険料率

print(f"健康保険料率（介護なし）：{kenko_rate*100:.2f}%")
print(f"健康保険料率（介護あり）：{kenko_kaigo*100:.2f}%")
print(f"厚生年金保険料率　　　　：{nenkin_rate*100:.2f}%")
```

```python
import time

# 47都道府県の料率を一括取得
prefs = [
    "北海道", "青森", "岩手", "宮城", "秋田", "山形", "福島",
    "茨城", "栃木", "群馬", "埼玉", "千葉", "東京", "神奈川",
    "新潟", "富山", "石川", "福井", "山梨", "長野", "岐阜",
    "静岡", "愛知", "三重", "滋賀", "京都", "大阪", "兵庫",
    "奈良", "和歌山", "鳥取", "島根", "岡山", "広島", "山口",
    "徳島", "香川", "愛媛", "高知", "福岡", "佐賀", "長崎",
    "熊本", "大分", "宮崎", "鹿児島", "沖縄"
]

# 料率辞書を作成
hoken_rates = {}

for pref in prefs:
    try:
        df = df_raw[pref]
        rate_row = df.iloc[6]

        kenko      = float(rate_row.iloc[5])   # 健康保険料率（介護なし）
        kenko_kaigo = float(rate_row.iloc[7])  # 健康保険料率（介護あり）
        nenkin     = float(rate_row.iloc[11])  # 厚生年金保険料率

        hoken_rates[pref] = {
            "健康保険料率":      kenko,
            "健康保険料率_介護": kenko_kaigo,
            "厚生年金保険料率":  nenkin
        }
        print(f"✅ {pref}：健康{kenko*100:.2f}% / 介護あり{kenko_kaigo*100:.2f}%")

    except Exception as e:
        print(f"⚠️ {pref}：エラー → {e}")

print(f"\n取得完了：{len(hoken_rates)}都道府県")
```

```python
from decimal import Decimal, ROUND_HALF_UP
import math

# ============================================================
# 令和8年度 追加料率
# ============================================================
SHIEN_KIN_RATE = 0.0023   # 子ども・子育て支援金（0.23%）
KAIGO_RATE     = 0.0162   # 介護保険料率（1.62%）
KOYO_RATE      = 0.006    # 雇用保険料率（一般）

def round_shakai(yen):
    """社会保険の四捨五入（50銭以上切り上げ）"""
    return int(Decimal(str(yen)).quantize(Decimal('1'), rounding=ROUND_HALF_UP))

def floor_shakai(yen):
    """雇用保険の切り捨て"""
    return math.floor(yen)

def calc_all_hoken(row, shien_kin=True):
    """
    全社会保険料を計算する
    shien_kin=True → 子ども・子育て支援金を含める（5月支払い分以降）
    shien_kin=False → 支援金を含めない（4月支払い分まで）
    """
    kyuyo  = row["標準報酬月額"]
    age    = row["年齢"]
    pref   = row["都道府県"]
    gyoshu = row.get("業種", "一般")

    rates = hoken_rates.get(pref, hoken_rates["東京"])

    # 40歳以上かどうかで料率を切り替え
    if age >= 40:
        kenko_rate = rates["健康保険料率_介護"]
    else:
        kenko_rate = rates["健康保険料率"]

    # ①健康保険料（四捨五入）
    kenko = round_shakai(kyuyo * kenko_rate / 2)

    # ②子ども・子育て支援金（四捨五入）
    shien = round_shakai(kyuyo * SHIEN_KIN_RATE / 2) if shien_kin else 0

    # ③厚生年金保険料（四捨五入）
    nenkin = round_shakai(kyuyo * rates["厚生年金保険料率"] / 2)

    # ④雇用保険料（切り捨て）
    koyo = floor_shakai(kyuyo * KOYO_RATE)

    # 合計
    total = kenko + shien + nenkin + koyo

    return pd.Series({
        "健康保険料":       kenko,
        "子育て支援金":     shien,
        "厚生年金料":       nenkin,
        "雇用保険料":       koyo,
        "社保控除合計":     total,
        "手取り概算":       kyuyo - total
    })

# テスト用従業員データ
employees = pd.DataFrame({
    "氏名":         ["田中 太郎", "鈴木 花子", "佐藤 次郎", "山田 美咲", "伊藤 健一"],
    "年齢":         [45, 38, 52, 29, 41],
    "都道府県":     ["東京", "神奈川", "大阪", "千葉", "東京"],
    "標準報酬月額": [300000, 250000, 400000, 220000, 350000],
    "業種":         ["一般", "一般", "一般", "一般", "一般"]
})

# 支援金あり（5月支払い分以降）
result_may = employees.join(
    employees.apply(lambda r: calc_all_hoken(r, shien_kin=True), axis=1)
)

# 支援金なし（4月支払い分まで）
result_apr = employees.join(
    employees.apply(lambda r: calc_all_hoken(r, shien_kin=False), axis=1)
)

print("【4月支払い分まで（支援金なし）】")
print(result_apr[["氏名", "健康保険料", "子育て支援金", "厚生年金料", "雇用保険料", "社保控除合計"]])
print()
print("【5月支払い分以降（支援金あり）】")
print(result_may[["氏名", "健康保険料", "子育て支援金", "厚生年金料", "雇用保険料", "社保控除合計"]])
```
